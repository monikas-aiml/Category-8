```python
# Medical Diagnosis Using Forward Chaining in Python

# Step 1: Define medical rules
rules = [
    ({"fever", "cough", "body_pain"}, "Possible Flu"),
    ({"sneezing", "runny_nose", "itchy_eyes"}, "Possible Allergy"),
    ({"sore_throat", "runny_nose", "cough"}, "Possible Common Cold"),
    ({"fever", "cough", "sore_throat"}, "Possible Respiratory Infection"),
    ({"fever", "headache", "body_pain"}, "Needs Further Evaluation")
]

# Step 2: Get symptoms from the user
print("MEDICAL DIAGNOSIS SYSTEM")
print("Enter symptoms separated by commas.")
print("Available symptoms: fever, cough, body_pain, headache,")
print("sore_throat, sneezing, runny_nose, itchy_eyes")

user_input = input("\nEnter your symptoms: ")

# Convert input into facts
facts = {
    symptom.strip().lower().replace(" ", "_")
    for symptom in user_input.split(",")
    if symptom.strip()
}

# Step 3: Forward chaining algorithm
def forward_chaining(facts, rules):
    conclusions = set()
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if conditions.issubset(facts):
                if conclusion not in facts:
                    facts.add(conclusion)
                    conclusions.add(conclusion)
                    changed = True
                    print("Rule Applied:", conclusion)

    return conclusions

# Step 4: Generate possible conclusions
results = forward_chaining(facts, rules)

# Step 5: Display the result
print("\n--- DIAGNOSIS RESULTS ---")

if results:
    for result in sorted(results):
        print(result)
else:
    print("No matching rule found.")
    print("Further medical evaluation may be needed.")
```

```python
 #Autonomous Vehicle Using Forward Chaining

# Step 1: Define Facts
facts = {
    "traffic_light_green",
    "road_clear",
    "safe_to_move"
}

print("Initial Facts:", facts)

# Step 2: Define Rules
rules = [
    ({"obstacle_ahead"}, "stop"),
    ({"traffic_light_red"}, "stop"),
    ({"traffic_light_green", "road_clear"}, "move_forward"),
    ({"obstacle_left", "road_clear"}, "turn_right"),
    ({"obstacle_right", "road_clear"}, "turn_left")
]

# Step 3: Forward Chaining Algorithm
def forward_chaining(facts, rules):
    conclusions = set()
    changed = True

    while changed:
        changed = False

        for conditions, action in rules:
            if conditions.issubset(facts):
                if action not in facts:
                    facts.add(action)
                    conclusions.add(action)
                    changed = True
                    print("Rule matched:", conditions)
                    print("Conclusion:", action)

    return conclusions

# Step 4: Apply Forward Chaining
results = forward_chaining(facts, rules)

# Step 5: Display Final Result
print("\nFinal Facts:", facts)
print("\nVehicle Decision:")

if "stop" in results:
    print("STOP")
elif "turn_left" in results:
    print("TURN LEFT")
elif "turn_right" in results:
    print("TURN RIGHT")
elif "move_forward" in results:
    print("MOVE FORWARD")
else:
    print("STOP - No safe decision found")

print("\nEducational simulation only.")
```

```python

# Industrial Manufacturing Using Forward Chaining

# Step 1: Define initial facts
facts = {
    "raw_material_available",
    "machine_operational",
    "worker_available",
    "quality_check_passed"
}

print("Initial Facts:")
for fact in sorted(facts):
    print("-", fact)

# Step 2: Define IF-THEN rules
rules = [
    ({"raw_material_available"}, "material_ready"),
    ({"machine_operational"}, "machine_ready"),
    ({"worker_available"}, "staff_ready"),
    ({"material_ready", "machine_ready", "staff_ready"},
     "production_can_start"),
    ({"production_can_start"}, "product_manufactured"),
    ({"product_manufactured", "quality_check_passed"},
     "product_approved"),
    ({"product_manufactured", "quality_check_failed"},
     "product_rejected")
]

# Step 3: Forward chaining algorithm
def forward_chaining(facts, rules):
    conclusions = set()
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if conditions.issubset(facts):
                if conclusion not in facts:
                    facts.add(conclusion)
                    conclusions.add(conclusion)
                    changed = True

                    print("\nRule matched:", conditions)
                    print("New fact:", conclusion)

    return conclusions

# Step 4: Apply forward chaining
print("\nApplying Forward Chaining...")
results = forward_chaining(facts, rules)

# Step 5: Display final result
print("\n--- MANUFACTURING RESULTS ---")

if "production_can_start" in facts:
    print("Production can start.")

if "product_manufactured" in facts:
    print("Product manufactured.")

if "product_approved" in facts:
    print("Product approved after quality check.")

if "product_rejected" in facts:
    print("Product rejected. Further inspection required.")

if "production_can_start" not in facts:
    print("Production cannot start.")
    print("Check raw materials, machine, and worker availability.")

print("\nFinal Facts:")
for fact in sorted(facts):
    print("-", fact)
```
